# **Tokenization**

Example:

---


    1- hello my name is abdullah khamis
    ['hello', 'my', 'name', 'is', 'abdullah', 'khamis'] ---> word tokenization
    2- hello, my name is abdullah khamis, and i'm Ai Engineer
    ['hello', 'my name is abdullah khamis', 'and i'm Ai Engineer'] ---> sent tokenization
    3- hello my name is abdullah khamis
    ['hell', 'o my', ' nam', 'e is', ' abd', 'ulla', 'h kha', 'mis'] ---> chunking

```python
'''
tokenization is where text broken down into smaller units called tokens.
tokens can be words, subwords, characters, or even sentences, depending on the approach.
it is converting unstructured text into smaller chunks makes it easier to analyze.
'''

# word tokenization
'''
split text at word boundaries producing individual words or tokens
Example_1: "my cat color is black" ---> ['my', 'cat', 'color', 'is', 'black']
Example_2: "mind cloud is a robotics team" ---> ['mind', 'cloud', 'is', 'a', 'robotics', 'team']
'''

import nltk
from nltk import word_tokenize
# nltk.download('punkt_tab')

text = "my name is abdullah khamis"
tokens = word_tokenize(text)
print(tokens)
```

```python
'''
word tokenization is limited for language where word boundaries aren't defined by space (chinese - Japanese)
'''
```

```python
# sentence tokenization
'''
split text at sentence boundaries, treating each sentence as a tokens
'''

from nltk import sent_tokenize
text = 'NLP is facsinating. it has many applications.'
tokens = sent_tokenize(text)
print(tokens)
```

# **Stop Words**

```python
'''
Stop words are words in a stoplist which are filtered out before or after processing language data because they are insignificant.

In English, common stop words include:
a - an - the - on - in - at - by - in - but - or - he - she - it - they - is - are - was - were

In German, common stop words include:
    der - die - das - und - in - zu - mit - auf - von - ist - du - ich - es - nicht - aber
'''

from nltk.corpus import stopwords
# nltk.download('stopwords')

print("English Stop Words: ", stopwords.words('english'))
print("German Stop Words: ", stopwords.words('german'))
print("Arabic Stop Words: ", stopwords.words('arabic'))
```

```python
stopwords = set(stopwords.words('english'))

text = 'my name is abdullah khamis, i like study NLP, and my age is 20 year'
tokens = word_tokenize(text)

filtered = [i for i in tokens if i not in stopwords]
filtered = " ".join(filtered)

print(filtered)
```

# **Stemming**

```python
'''
technique that reduce word to its base or root removing prefixes and suffixes.
the root form or stem obtained through stemming may not always be a valid word but serves as a representation for words with similar meanings in tasks

running → run
studies → studi
happier → happi
cats → cat
easily → easili
played → play
'''

# porter stemmer
'''
specifically designed for english language, it was introduced by martin porter in 1980 and operates using a set of rules tailored to the morphology of english words

prefixes

[un] unhappy
[re] replay
[in] incomplete
[dis] disagree
[im] impossible
[en] enable
[pre] preview
[mis] misunderstand
[non] nonprofit
[under] underestimate
'''

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ['Playing', 'cars', 'unhappy']
for i in words:
   print(i, ' | ', stemmer.stem(i))
```

```python
# snowball stemmer

'''
improved and more flexable version of porter stemmer, it support multibale language
'''

from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("german")
words = [
    "Häuser",
    "Häusern",
    "laufen",
    "läuft",
    "gelaufen",
    "Kinder",
    "Kindern",
    "schönen",
    "schönste",
    "Freundschaft",
    "Freundschaften"
]

for word in words:
    stemmed = stemmer.stem(word)
    print(f"{word} -> {stemmed}")
```

```python
# ISRI stemmer
'''
designed specifically for the Arabic language. It was developed by institute for arabic and islamic studies (ISRI)

ISRI is founded on modern standard arabic. It typically used with minimal adaptations to handle various Arabic dialects
such as Egyptian Arabic, or moroccan arabic

common arabic prefixes

ال <- الكتاب
و <- وكتاب
بال <- بالكتاب
لل <- للكتاب <- للكتابين
ما <- ما
وما <- وما <- وماهو
من <- من
'''
from nltk.stem import ISRIStemmer

isri = ISRIStemmer()
words = ['المجد', 'المكتبة', 'ماهو']
for i in words:
  print(i, ' --> ', isri.stem(i))
```

# **Lemmatization**

```python
import nltk
from nltk.stem import WordNetLemmatizer
# nltk.download("wordnet")
# nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

words = [
    "running",
    "studies",
    "better",
    "children",
    "cars",
    "played",
    "wolves"
]

for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(f"{word} -> {lemma}")
```

# **Spell Correction**

```python
'''
Spell correction is an NLP technique used to detect and correct misspelled words.
It compares an incorrect word with possible words in a dictionary and selects the most likely correct word.
Common approaches use edit distance, which measures how many insertions, deletions, substitutions, or transpositions are needed to transform one word into another.
'''

from spellchecker import SpellChecker

spell = SpellChecker()

text = "I am lerning machne lernning and natural langauge processing"

words = text.split()
corrected_words = []

for word in words:
    correction = spell.correction(word)
    corrected_words.append(correction)

corrected_text = " ".join(corrected_words)

print("Original:")
print(text)

print("\nCorrected:")
print(corrected_text)
```
